```python
from pyspark.sql.functions import col, sum
df = spark.table("my_catalog.my_db.order_line")

filtered_df = df.filter(
    (col("order_status") == "Delivered") |
    (col("order_status") == "Cancelled")
    )


grouped_df = filtered_df.groupBy("order_status").agg(
    sum("price").alias("total_price"),
    sum("quantity").alias("total_quantity")
)
display(grouped_df)

grouped_df.write.mode("overwrite").saveAsTable("my_catalog.my_db.transformed_table")
```
